## 1. Array və Dynamic Array

- **FixedArray:** sabit tutumlu massivdir.
- **DynamicArray:** tutum dolduqda yaddaş sahəsini artırır.
- Hər iki strukturda indekslə oxuma, əlavə etmə, silmə və axtarış əməliyyatları reallaşdırılıb.


```python
class FixedArray:
    """Tutumu dəyişməyən, indekslənən massiv."""
    def __init__(self, capacity: int):
        if capacity < 0:
            raise ValueError("Tutum mənfi ola bilməz.")
        self._data = [None] * capacity
        self._size = 0

    def __len__(self):
        return self._size

    @property
    def capacity(self):
        return len(self._data)

    def _check_index(self, index):
        if not 0 <= index < self._size:
            raise IndexError("İndeks sərhəddən kənardadır.")

    def append(self, value):
        if self._size >= self.capacity:
            raise OverflowError("FixedArray üçün boş yer qalmayıb.")
        self._data[self._size] = value
        self._size += 1

    def insert(self, index, value):
        if not 0 <= index <= self._size:
            raise IndexError("Əlavə etmə indeksi yanlışdır.")
        if self._size >= self.capacity:
            raise OverflowError("FixedArray üçün boş yer qalmayıb.")
        for i in range(self._size, index, -1):
            self._data[i] = self._data[i - 1]
        self._data[index] = value
        self._size += 1

    def delete(self, index):
        self._check_index(index)
        removed = self._data[index]
        for i in range(index, self._size - 1):
            self._data[i] = self._data[i + 1]
        self._size -= 1
        self._data[self._size] = None
        return removed

    def get(self, index):
        self._check_index(index)
        return self._data[index]

    def search(self, value):
        for i in range(self._size):
            if self._data[i] == value:
                return i
        return -1

    def __iter__(self):
        for i in range(self._size):
            yield self._data[i]

    def __repr__(self):
        return f"FixedArray({list(self)}, capacity={self.capacity})"


class DynamicArray:
    """Tutum dolduqda buferi ikiqat böyüyən dinamik massiv."""
    def __init__(self, initial_capacity=2):
        if initial_capacity < 1:
            initial_capacity = 1
        self._data = [None] * initial_capacity
        self._size = 0
        self.resize_count = 0

    def __len__(self):
        return self._size

    @property
    def capacity(self):
        return len(self._data)

    def _resize(self, new_capacity):
        new_data = [None] * new_capacity
        for i in range(self._size):
            new_data[i] = self._data[i]
        self._data = new_data
        self.resize_count += 1

    def _ensure_capacity(self, required):
        if required > self.capacity:
            new_capacity = max(required, self.capacity * 2)
            self._resize(new_capacity)

    def _check_index(self, index):
        if not 0 <= index < self._size:
            raise IndexError("İndeks sərhəddən kənardadır.")

    def append(self, value):
        self._ensure_capacity(self._size + 1)
        self._data[self._size] = value
        self._size += 1

    def insert(self, index, value):
        if not 0 <= index <= self._size:
            raise IndexError("Əlavə etmə indeksi yanlışdır.")
        self._ensure_capacity(self._size + 1)
        for i in range(self._size, index, -1):
            self._data[i] = self._data[i - 1]
        self._data[index] = value
        self._size += 1

    def delete(self, index):
        self._check_index(index)
        removed = self._data[index]
        for i in range(index, self._size - 1):
            self._data[i] = self._data[i + 1]
        self._size -= 1
        self._data[self._size] = None
        return removed

    def get(self, index):
        self._check_index(index)
        return self._data[index]

    def search(self, value):
        for i in range(self._size):
            if self._data[i] == value:
                return i
        return -1

    def index_of(self, value):
        return self.search(value)

    def __iter__(self):
        for i in range(self._size):
            yield self._data[i]

    def __repr__(self):
        return f"DynamicArray({list(self)}, capacity={self.capacity})"


# Kiçik nümunə
fixed_demo = FixedArray(5)
for item in [10, 20, 30]:
    fixed_demo.append(item)
fixed_demo.insert(1, 15)
print("FixedArray:", fixed_demo, "| 30-un indeksi:", fixed_demo.search(30))

dynamic_demo = DynamicArray(2)

for item in [10, 20, 30, 40]:
    dynamic_demo.append(item)
print("DynamicArray:", dynamic_demo, "| tutum:", dynamic_demo.capacity,
      "| resize sayı:", dynamic_demo.resize_count)
```

## 2. Singly və Doubly Linked List

Singly Linked List hər düyündə növbəti düyünü, Doubly Linked List isə həm əvvəlki, həm də növbəti düyünü saxlayır. Hər iki strukturda sona əlavə etmə `O(1)`-dir; sondan silmə isə Singly Linked List-də `O(n)`, Doubly Linked List-də `O(1)`-dir.


```python
class _SNode:
    def __init__(self, value):
        self.value = value
        self.next = None


class SinglyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
        self._size = 0

    def __len__(self):
        return self._size

    def is_empty(self):
        return self._size == 0

    def prepend(self, value):
        node = _SNode(value)
        node.next = self.head
        self.head = node
        if self.tail is None:
            self.tail = node
        self._size += 1
        return node

    def append(self, value):
        node = _SNode(value)
        if self.tail is None:
            self.head = self.tail = node
        else:
            self.tail.next = node
            self.tail = node
        self._size += 1
        return node

    def popleft(self):
        if self.head is None:
            raise IndexError("Siyahı boşdur.")
        value = self.head.value
        self.head = self.head.next
        self._size -= 1
        if self.head is None:
            self.tail = None
        return value

    def pop_back(self):
        if self.head is None:
            raise IndexError("Siyahı boşdur.")
        if self.head is self.tail:
            value = self.head.value
            self.head = self.tail = None
            self._size = 0
            return value
        current = self.head
        while current.next is not self.tail:
            current = current.next
        value = self.tail.value
        current.next = None
        self.tail = current
        self._size -= 1
        return value

    def find(self, value):
        current = self.head
        while current is not None:
            if current.value == value:
                return current
            current = current.next
        return None

    def remove(self, value):
        previous = None
        current = self.head
        while current is not None:
            if current.value == value:
                if previous is None:
                    self.head = current.next
                else:
                    previous.next = current.next
                if current is self.tail:
                    self.tail = previous
                self._size -= 1
                return True
            previous, current = current, current.next
        return False

    def __iter__(self):
        current = self.head
        while current is not None:
            yield current.value
            current = current.next


class _DNode:
    def __init__(self, value):
        self.value = value
        self.prev = None
        self.next = None


class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
        self._size = 0

    def __len__(self):
        return self._size

    def is_empty(self):
        return self._size == 0

    def prepend(self, value):
        node = _DNode(value)
        node.next = self.head
        if self.head is not None:
            self.head.prev = node
        else:
            self.tail = node
        self.head = node
        self._size += 1
        return node

    def append(self, value):
        node = _DNode(value)
        node.prev = self.tail
        if self.tail is not None:
            self.tail.next = node
        else:
            self.head = node
        self.tail = node
        self._size += 1
        return node

    def remove_node(self, node):
        if node.prev is not None:
            node.prev.next = node.next
        else:
            self.head = node.next
        if node.next is not None:
            node.next.prev = node.prev
        else:
            self.tail = node.prev
        self._size -= 1
        return node.value

    def popleft(self):
        if self.head is None:
            raise IndexError("Siyahı boşdur.")
        return self.remove_node(self.head)

    def pop_back(self):
        if self.tail is None:
            raise IndexError("Siyahı boşdur.")
        return self.remove_node(self.tail)

    def find_node(self, value):
        current = self.head
        while current is not None:
            if current.value == value:
                return current
            current = current.next
        return None

    def find(self, value):
        node = self.find_node(value)
        return node.value if node else None

    def remove(self, value):
        node = self.find_node(value)
        if node is None:
            return False
        self.remove_node(node)
        return True

    def __iter__(self):
        current = self.head
        while current is not None:
            yield current.value
            current = current.next


sll_demo = SinglyLinkedList()
for x in [1, 2, 3]:
    sll_demo.append(x)
dll_demo = DoublyLinkedList()
for x in [1, 2, 3]:
    dll_demo.append(x)
print("Singly:", list(sll_demo), "| pop_back:", sll_demo.pop_back())
print("Doubly:", list(dll_demo), "| pop_back:", dll_demo.pop_back())
```

## 3. Stack, Queue və Deque

- **Stack (LIFO):** son kataloq əməliyyatını geri qaytarır.
- **Queue (FIFO):** kitab gözləyən oxucuları növbə ilə idarə edir.
- **Deque:** təcili sorğuları növbənin əvvəlinə, adi sorğuları sonuna əlavə edir.

Strukturlar ayrıca implementasiya olunub.


```python
class Stack:
    def __init__(self):
        self._items = SinglyLinkedList()

    def push(self, value):
        self._items.prepend(value)

    def pop(self):
        return self._items.popleft()

    def peek(self):
        if self._items.is_empty():
            raise IndexError("Stack boşdur.")
        return self._items.head.value

    def is_empty(self):
        return self._items.is_empty()

    def __len__(self):
        return len(self._items)


class Queue:
    def __init__(self):
        self._items = SinglyLinkedList()

    def enqueue(self, value):
        self._items.append(value)

    def dequeue(self):
        return self._items.popleft()

    def peek(self):
        if self._items.is_empty():
            raise IndexError("Queue boşdur.")
        return self._items.head.value

    def is_empty(self):
        return self._items.is_empty()

    def __len__(self):
        return len(self._items)


class Deque:
    def __init__(self):
        self._items = DoublyLinkedList()

    def add_front(self, value):
        self._items.prepend(value)

    def add_rear(self, value):
        self._items.append(value)

    def remove_front(self):
        return self._items.popleft()

    def remove_rear(self):
        return self._items.pop_back()

    def is_empty(self):
        return self._items.is_empty()

    def __len__(self):
        return len(self._items)

    def __iter__(self):
        return iter(self._items)


stack_demo = Stack()
stack_demo.push("A")
stack_demo.push("B")
queue_demo = Queue()
queue_demo.enqueue("Oxucu 1")
queue_demo.enqueue("Oxucu 2")
deque_demo = Deque()
deque_demo.add_rear("Adi sorğu")
deque_demo.add_front("Təcili sorğu")
print("Stack pop:", stack_demo.pop())
print("Queue dequeue:", queue_demo.dequeue())
print("Deque ardıcıllığı:", list(deque_demo))
```

## 4. Chaining üsullu Custom Hash Table

Açar–dəyər cütləri saxlanılır. Eyni indeksə düşən açarlar collision yaradır və linked chain vasitəsilə idarə olunur. Yüklənmə əmsalı `0.75`-i keçdikdə cədvəl böyüdülür.


```python
class _HashNode:
    def __init__(self, key, value, next_node=None):
        self.key = key
        self.value = value
        self.next = next_node


class ChainingHashTable:
    def __init__(self, capacity=8, max_load_factor=0.75):
        self._capacity = max(2, capacity)
        self._buckets = [None] * self._capacity
        self._size = 0
        self.max_load_factor = max_load_factor
        self.collision_events = 0

    def __len__(self):
        return self._size

    @property
    def capacity(self):
        return self._capacity

    @property
    def load_factor(self):
        return self._size / self._capacity

    def _index(self, key):
        # Deterministik polynomial hash; müxtəlif tipli açarlar str-ə çevrilir.
        text = str(key)
        hash_value = 0
        for char in text:
            hash_value = (hash_value * 31 + ord(char)) % self._capacity
        return hash_value

    def _find_node(self, key):
        index = self._index(key)
        current = self._buckets[index]
        while current is not None:
            if current.key == key:
                return current
            current = current.next
        return None

    def put(self, key, value):
        index = self._index(key)
        current = self._buckets[index]
        while current is not None:
            if current.key == key:
                current.value = value
                return False  # Mövcud açar yeniləndi; ölçü dəyişmədi.
            current = current.next

        if self._buckets[index] is not None:
            self.collision_events += 1
        self._buckets[index] = _HashNode(key, value, self._buckets[index])
        self._size += 1
        if self.load_factor > self.max_load_factor:
            self._resize(self._capacity * 2)
        return True

    def get(self, key, default=None):
        node = self._find_node(key)
        return node.value if node is not None else default

    def contains(self, key):
        return self._find_node(key) is not None

    def remove(self, key):
        index = self._index(key)
        previous = None
        current = self._buckets[index]
        while current is not None:
            if current.key == key:
                if previous is None:
                    self._buckets[index] = current.next
                else:
                    previous.next = current.next
                self._size -= 1
                return current.value
            previous, current = current, current.next
        return None

    def _resize(self, new_capacity):
        old_buckets = self._buckets
        self._capacity = new_capacity
        self._buckets = [None] * self._capacity
        for head in old_buckets:
            current = head
            while current is not None:
                next_node = current.next
                index = self._index(current.key)
                current.next = self._buckets[index]
                self._buckets[index] = current
                current = next_node

    def max_chain_length(self):
        longest = 0
        for head in self._buckets:
            length = 0
            current = head
            while current is not None:
                length += 1
                current = current.next
            longest = max(longest, length)
        return longest

    def items(self):
        for head in self._buckets:
            current = head
            while current is not None:
                yield current.key, current.value
                current = current.next


# Collision nümayişi: "A" və "C" eyni bucket-ə düşür (capacity=2).
collision_demo = ChainingHashTable(capacity=2)
collision_demo.put("A", "Kitab A")
collision_demo.put("C", "Kitab C")
print("Açarlar:", list(collision_demo.items()))
print("Collision hadisələri:", collision_demo.collision_events)
print("Açar C üzrə axtarış:", collision_demo.get("C"))
```

## 5. Binary Search Tree (BST)

Sol alt ağacdakı açarlar düyündən kiçik, sağ alt ağacdakı açarlar isə böyükdür. Əlavə etmə, axtarış və silmə orta halda `O(log n)`, ən pis halda `O(n)` ola bilər.

Benchmark-da random və sorted insertion müqayisə edilir. Sıralanmış daxil etmə adi BST-ni balanssızlaşdıraraq zəncirə çevirə bilər.


```python
class _BSTNode:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None


class BinarySearchTree:
    def __init__(self):
        self.root = None
        self._size = 0

    def __len__(self):
        return self._size

    def insert(self, key):
        new_node = _BSTNode(key)
        if self.root is None:
            self.root = new_node
            self._size = 1
            return True
        current = self.root
        while True:
            if key == current.key:
                return False
            if key < current.key:
                if current.left is None:
                    current.left = new_node
                    self._size += 1
                    return True
                current = current.left
            else:
                if current.right is None:
                    current.right = new_node
                    self._size += 1
                    return True
                current = current.right

    def contains(self, key):
        current = self.root
        while current is not None:
            if key == current.key:
                return True
            current = current.left if key < current.key else current.right
        return False

    def delete(self, key):
        parent = None
        node = self.root
        while node is not None and node.key != key:
            parent = node
            node = node.left if key < node.key else node.right
        if node is None:
            return False

        # İki övladı varsa, sağ alt ağacın ən kiçik düyünü ilə əvəz edilir.
        if node.left is not None and node.right is not None:
            successor_parent = node
            successor = node.right
            while successor.left is not None:
                successor_parent = successor
                successor = successor.left
            node.key = successor.key
            parent = successor_parent
            node = successor

        child = node.left if node.left is not None else node.right
        if parent is None:
            self.root = child
        elif parent.left is node:
            parent.left = child
        else:
            parent.right = child
        self._size -= 1
        return True

    def inorder(self):
        stack = []
        current = self.root
        while stack or current is not None:
            while current is not None:
                stack.append(current)
                current = current.left
            current = stack.pop()
            yield current.key
            current = current.right

    def height(self):
        if self.root is None:
            return 0
        maximum = 0
        stack = [(self.root, 1)]
        while stack:
            node, depth = stack.pop()
            maximum = max(maximum, depth)
            if node.left is not None:
                stack.append((node.left, depth + 1))
            if node.right is not None:
                stack.append((node.right, depth + 1))
        return maximum


bst_demo = BinarySearchTree()
for value in [50, 30, 70, 20, 40, 60, 80]:
    bst_demo.insert(value)
print("BST inorder:", list(bst_demo.inorder()))
print("60 mövcuddur?", bst_demo.contains(60), "| hündürlük:", bst_demo.height())
bst_demo.delete(50)
print("50 silindikdən sonra:", list(bst_demo.inorder()))
```

## 6. Avtomatik testlər

Əsas əməliyyatlar və boş struktur, collision, BST silinməsi kimi kənar hallar yoxlanılır.


```python
def run_all_tests():
    # FixedArray və DynamicArray
    a = FixedArray(5)
    for x in [1, 3, 4]:
        a.append(x)
    a.insert(1, 2)
    assert list(a) == [1, 2, 3, 4]
    assert a.search(3) == 2
    assert a.delete(1) == 2
    assert list(a) == [1, 3, 4]

    da = DynamicArray(1)
    for x in range(20):
        da.append(x)
    assert len(da) == 20 and da.capacity >= 20
    assert da.search(19) == 19
    assert da.delete(0) == 0

    # Singly və Doubly Linked List
    sl = SinglyLinkedList()
    sl.append(1); sl.append(2); sl.prepend(0)
    assert list(sl) == [0, 1, 2]
    assert sl.remove(1) and sl.pop_back() == 2 and sl.popleft() == 0
    assert sl.is_empty()

    dl = DoublyLinkedList()
    dl.append(1); dl.append(2); dl.prepend(0)
    assert list(dl) == [0, 1, 2]
    assert dl.pop_back() == 2 and dl.popleft() == 0 and dl.remove(1)
    assert dl.is_empty()

    # Stack, Queue, Deque
    st = Stack(); st.push(1); st.push(2)
    assert st.peek() == 2 and st.pop() == 2 and st.pop() == 1
    q = Queue(); q.enqueue("a"); q.enqueue("b")
    assert q.dequeue() == "a" and q.dequeue() == "b"
    dq = Deque(); dq.add_rear(2); dq.add_front(1); dq.add_rear(3)
    assert dq.remove_front() == 1 and dq.remove_rear() == 3 and dq.remove_front() == 2

    # Hash Table collision və silmə
    ht = ChainingHashTable(capacity=2)
    ht.put("A", 1); ht.put("C", 3)
    assert ht.get("A") == 1 and ht.get("C") == 3
    assert ht.collision_events >= 1
    assert ht.remove("A") == 1 and not ht.contains("A")
    for i in range(30):
        ht.put(f"key-{i}", i)
    assert ht.get("key-29") == 29

    # BST: yarpaq, bir övlad və iki övladın silinməsi
    tree = BinarySearchTree()
    for x in [50, 30, 70, 20, 40, 60, 80, 65]:
        tree.insert(x)
    assert tree.contains(50) and not tree.contains(999)
    assert tree.delete(20)  # yarpaq
    assert tree.delete(60)  # bir övlad
    assert tree.delete(50)  # iki övlad
    assert list(tree.inorder()) == [30, 40, 65, 70, 80]

    # Kitabxana inteqrasiyası: waitlist, rezerv və Stack ilə undo
    system = LibraryManagementSystem()
    assert system.add_book(101, "Test kitabı", "Müəllif", 2024)
    assert system.borrow_book(101, "Əli").startswith("Kitab Əli")
    assert system.borrow_book(101, "Aysel").startswith("Kitab hazırda")
    assert "Aysel" in system.return_book(101)
    assert system.borrow_book(101, "Aysel").startswith("Kitab Aysel")
    assert system.return_book(101).startswith("Kitab uğurla")
    assert system.add_book(102, "Undo testi", "Müəllif", 2023)
    assert system.undo_last_catalog_change().startswith("Geri alındı")
    assert system.get_book(102) is None

    print("✅ Bütün avtomatik testlər uğurla keçdi.")
```

## 7. Performans benchmark-ları

Ölçmələr `time.perf_counter_ns()` ilə aparılır; nəticə bir neçə təkrarın medianıdır. Vaxtlar Colab runtime-ının yükündən asılı olaraq dəyişə bilər.


```python
def median_operation_ms(factory, operation, repeats=5):
    samples = []
    for _ in range(repeats):
        structure = factory()  # hazırlıq ölçülən vaxta daxil deyil
        start = time.perf_counter_ns()
        operation(structure)
        samples.append((time.perf_counter_ns() - start) / 1_000_000)
    return statistics.median(samples)


def build_fixed(values, capacity):
    structure = FixedArray(capacity)
    for value in values:
        structure.append(value)
    return structure


def build_dynamic(values, capacity=None):
    structure = DynamicArray(capacity if capacity is not None else 2)
    for value in values:
        structure.append(value)
    return structure


# 7.1 Fixed Array və Dynamic Array
array_sizes = [100, 500, 1000, 3000, 5000]
array_rows = []
repeats = 5

for n in array_sizes:
    values = list(range(n))
    middle = n // 2
    fixed_factory = lambda values=values, n=n: build_fixed(values, n + 1)
    dynamic_factory = lambda values=values, n=n: build_dynamic(values, n + 1)

    for label, factory in [("Fixed Array", fixed_factory), ("Dynamic Array", dynamic_factory)]:
        insert_ms = median_operation_ms(factory, lambda s, m=middle: s.insert(m, -1), repeats)
        delete_ms = median_operation_ms(factory, lambda s, m=middle: s.delete(m), repeats)
        search_ms = median_operation_ms(factory, lambda s, target=n-1: s.search(target), repeats)
        array_rows.extend([
            {"Ölçü": n, "Struktur": label, "Əməliyyat": "Middle insert", "Vaxt (ms)": insert_ms},
            {"Ölçü": n, "Struktur": label, "Əməliyyat": "Middle delete", "Vaxt (ms)": delete_ms},
            {"Ölçü": n, "Struktur": label, "Əməliyyat": "Search (last)", "Vaxt (ms)": search_ms},
        ])

    # Batch append: orta bir əlavə etmə vaxtı. DynamicArray n tutumundan başlayır.
    batch = max(10, n // 10)
    fixed_append_factory = lambda values=values, n=n, batch=batch: build_fixed(values, n + batch)
    dynamic_append_factory = lambda values=values, n=n: build_dynamic(values, n)
    fixed_batch_ms = median_operation_ms(
        fixed_append_factory,
        lambda s, n=n, batch=batch: [s.append(n + i) for i in range(batch)], repeats
    ) / batch
    dynamic_batch_ms = median_operation_ms(
        dynamic_append_factory,
        lambda s, n=n, batch=batch: [s.append(n + i) for i in range(batch)], repeats
    ) / batch
    array_rows.append({"Ölçü": n, "Struktur": "Fixed Array", "Əməliyyat": "Batch append (orta)", "Vaxt (ms)": fixed_batch_ms})
    array_rows.append({"Ölçü": n, "Struktur": "Dynamic Array", "Əməliyyat": "Batch append (orta)", "Vaxt (ms)": dynamic_batch_ms})

array_results = pd.DataFrame(array_rows)
print("ARRAY BENCHMARK NƏTİCƏLƏRİ (ms)")
display(array_results.round(6))

for operation in ["Middle insert", "Middle delete", "Search (last)", "Batch append (orta)"]:
    subset = array_results[array_results["Əməliyyat"] == operation]
    plt.figure(figsize=(8, 4.5))
    for structure_name in ["Fixed Array", "Dynamic Array"]:
        part = subset[subset["Struktur"] == structure_name]
        plt.plot(part["Ölçü"], part["Vaxt (ms)"], marker="o", label=structure_name)
    plt.xscale("log")
    plt.yscale("log")
    plt.xlabel("Elementlərin sayı (log scale)")
    plt.ylabel("Median vaxt (ms, log scale)")
    plt.title(f"Array müqayisəsi — {operation}")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.show()
```

```python
# 7.2 Singly və Doubly Linked List
linked_sizes = [100, 500, 1000, 3000, 5000]
linked_rows = []

for n in linked_sizes:
    values = list(range(n))
    for label, list_class in [("Singly Linked List", SinglyLinkedList), ("Doubly Linked List", DoublyLinkedList)]:
        def factory(list_class=list_class, values=values):
            structure = list_class()
            for value in values:
                structure.append(value)
            return structure

        append_ms = median_operation_ms(factory, lambda s, n=n: s.append(n), repeats)
        search_ms = median_operation_ms(factory, lambda s, target=n-1: s.find(target), repeats)
        pop_back_ms = median_operation_ms(factory, lambda s: s.pop_back(), repeats)
        linked_rows.extend([
            {"Ölçü": n, "Struktur": label, "Əməliyyat": "Append", "Vaxt (ms)": append_ms},
            {"Ölçü": n, "Struktur": label, "Əməliyyat": "Search (last)", "Vaxt (ms)": search_ms},
            {"Ölçü": n, "Struktur": label, "Əməliyyat": "Pop back", "Vaxt (ms)": pop_back_ms},
        ])

linked_results = pd.DataFrame(linked_rows)
print("LINKED LIST BENCHMARK NƏTİCƏLƏRİ (ms)")
display(linked_results.round(6))

for operation in ["Append", "Search (last)", "Pop back"]:
    subset = linked_results[linked_results["Əməliyyat"] == operation]
    plt.figure(figsize=(8, 4.5))
    for structure_name in ["Singly Linked List", "Doubly Linked List"]:
        part = subset[subset["Struktur"] == structure_name]
        plt.plot(part["Ölçü"], part["Vaxt (ms)"], marker="o", label=structure_name)
    plt.xscale("log")
    plt.yscale("log")
    plt.xlabel("Düyünlərin sayı (log scale)")
    plt.ylabel("Median vaxt (ms, log scale)")
    plt.title(f"Linked List müqayisəsi — {operation}")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.show()
```

```python
# 7.3 Binary Search Tree: random və sorted insertion
bst_sizes = [100, 250, 500, 750, 1000]
bst_rows = []

for n in bst_sizes:
    sorted_values = list(range(n))
    random_values = sorted_values.copy()
    random.Random(100 + n).shuffle(random_values)

    def build_tree(values):
        tree = BinarySearchTree()
        for value in values:
            tree.insert(value)
        return tree

    for label, values in [("Random insertion", random_values), ("Sorted insertion", sorted_values)]:
        build_samples = []
        search_samples = []
        heights = []
        for _ in range(3):
            start = time.perf_counter_ns()
            tree = build_tree(values)
            build_samples.append((time.perf_counter_ns() - start) / 1_000_000)
            heights.append(tree.height())
            start = time.perf_counter_ns()
            tree.contains(n - 1)
            search_samples.append((time.perf_counter_ns() - start) / 1_000_000)
        bst_rows.append({
            "Ölçü": n,
            "Ssenari": label,
            "Build vaxtı (ms)": statistics.median(build_samples),
            "Search vaxtı (ms)": statistics.median(search_samples),
            "Ağac hündürlüyü": statistics.median(heights),
        })

bst_results = pd.DataFrame(bst_rows)
print("BST BENCHMARK NƏTİCƏLƏRİ")
display(bst_results.round(6))

plt.figure(figsize=(8, 4.5))
for scenario in ["Random insertion", "Sorted insertion"]:
    part = bst_results[bst_results["Ssenari"] == scenario]
    plt.plot(part["Ölçü"], part["Ağac hündürlüyü"], marker="o", label=scenario)
plt.xlabel("Düyünlərin sayı")
plt.ylabel("Ağacın hündürlüyü")
plt.title("BST: əlavə etmə ardıcıllığının ağac hündürlüyünə təsiri")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()

plt.figure(figsize=(8, 4.5))
for scenario in ["Random insertion", "Sorted insertion"]:
    part = bst_results[bst_results["Ssenari"] == scenario]
    plt.plot(part["Ölçü"], part["Search vaxtı (ms)"], marker="o", label=scenario)
plt.xlabel("Düyünlərin sayı")
plt.ylabel("Search vaxtı (ms)")
plt.title("BST: random və sorted insertion sonrası axtarış")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

## 8. Mini Library Management System

| Struktur | Tətbiqdəki vəzifə |
|---|---|
| Dynamic Array | Kitab kataloqu |
| Singly Linked List | Borc tarixçəsi |
| Doubly Linked List | Son baxılmış kitablar |
| Stack | Son kataloq əməliyyatını geri qaytarmaq |
| Queue | Kitab gözləyən oxucular |
| Deque | Xidmət sorğuları |
| Custom Hash Table | Kitab ID-si ilə axtarış və növbələrin saxlanması |
| BST | Kitab ID-lərinin sıralanması və axtarışı |


```python
@dataclass
class Book:
    book_id: int
    title: str
    author: str
    year: int
    is_available: bool = True
    borrowed_by: Optional[str] = None
    reserved_for: Optional[str] = None

    def __str__(self):
        if self.borrowed_by:
            status = f"Borcda: {self.borrowed_by}"
        elif self.reserved_for:
            status = f"Rezerv: {self.reserved_for}"
        else:
            status = "Mövcuddur" if self.is_available else "Mövcud deyil"
        return f"ID: {self.book_id} | {self.title} | {self.author} | {self.year} | {status}"


class LibraryManagementSystem:
    def __init__(self):
        self.books = DynamicArray(4)
        self.books_by_id = ChainingHashTable(capacity=8)
        self.book_tree = BinarySearchTree()
        self.borrowing_history = SinglyLinkedList()
        self.recently_viewed = DoublyLinkedList()
        self.undo_stack = Stack()
        self.wait_queues = ChainingHashTable(capacity=4)
        self.service_requests = Deque()

    def get_book(self, book_id):
        return self.books_by_id.get(book_id)

    def _insert_book_indexes(self, book):
        self.books.append(book)
        self.books_by_id.put(book.book_id, book)
        self.book_tree.insert(book.book_id)

    def _remove_book_indexes(self, book_id):
        book = self.books_by_id.get(book_id)
        if book is None:
            return None
        index = self.books.index_of(book)
        if index >= 0:
            self.books.delete(index)
        self.books_by_id.remove(book_id)
        self.book_tree.delete(book_id)
        return book

    def add_book(self, book_id, title, author, year):
        if self.get_book(book_id) is not None:
            return False
        book = Book(int(book_id), title.strip(), author.strip(), int(year))
        self._insert_book_indexes(book)
        self.undo_stack.push(("UNDO_ADD", book.book_id))
        return True

    def remove_book(self, book_id):
        book = self.get_book(book_id)
        if book is None or not book.is_available:
            return False
        waiting = self.wait_queues.get(book_id)
        if waiting is not None and not waiting.is_empty():
            return False
        removed = self._remove_book_indexes(book_id)
        if removed is not None:
            self.undo_stack.push(("UNDO_REMOVE", removed))
            return True
        return False

    def undo_last_catalog_change(self):
        if self.undo_stack.is_empty():
            return "Geri alına bilən əməliyyat yoxdur."
        action = self.undo_stack.pop()
        if action[0] == "UNDO_ADD":
            book = self._remove_book_indexes(action[1])
            if book is None:
                return "Əməliyyatı geri almaq mümkün olmadı."
            return f"Geri alındı: '{book.title}' kataloqdan silindi."
        if action[0] == "UNDO_REMOVE":
            book = action[1]
            if self.get_book(book.book_id) is not None:
                return "Əməliyyatı geri almaq mümkün olmadı: ID artıq mövcuddur."
            self._insert_book_indexes(book)
            return f"Geri alındı: '{book.title}' kataloqa qaytarıldı."
        return "Naməlum əməliyyat."

    def search_book(self, book_id):
        book = self.get_book(book_id)
        if book is not None:
            recent_key = f"{book.book_id} - {book.title}"
            self.recently_viewed.remove(recent_key)
            self.recently_viewed.append(recent_key)
        return book

    def borrow_book(self, book_id, borrower):
        book = self.get_book(book_id)
        if book is None:
            return "Bu ID ilə kitab tapılmadı."
        borrower = borrower.strip()
        if not borrower:
            return "Oxucunun adı boş ola bilməz."

        # Kitab gözləyən oxucu üçün rezerv edilibsə, yalnız həmin oxucu götürə bilər.
        if book.reserved_for is not None:
            if book.reserved_for.casefold() != borrower.casefold():
                return f"Kitab {book.reserved_for} üçün rezerv edilib."
            book.reserved_for = None
            book.is_available = False
            book.borrowed_by = borrower
            self.borrowing_history.append(f"{borrower} kitabı götürdü: {book.title} (ID {book_id})")
            return f"Kitab {borrower} üçün təhvil verildi."

        if book.is_available:
            book.is_available = False
            book.borrowed_by = borrower
            self.borrowing_history.append(f"{borrower} kitabı götürdü: {book.title} (ID {book_id})")
            return f"Kitab {borrower} üçün təhvil verildi."

        queue = self.wait_queues.get(book_id)
        if queue is None:
            queue = Queue()
            self.wait_queues.put(book_id, queue)
        queue.enqueue(borrower)
        return f"Kitab hazırda başqasındadır. {borrower} gözləmə növbəsinə əlavə edildi."

    def return_book(self, book_id):
        book = self.get_book(book_id)
        if book is None:
            return "Bu ID ilə kitab tapılmadı."
        if book.borrowed_by is None:
            return "Bu kitab hazırda heç bir oxucuda deyil."
        previous_borrower = book.borrowed_by
        book.borrowed_by = None
        queue = self.wait_queues.get(book_id)
        if queue is not None and not queue.is_empty():
            next_reader = queue.dequeue()
            book.reserved_for = next_reader
            book.is_available = False
            self.borrowing_history.append(
                f"{previous_borrower} kitabı qaytardı: {book.title}; növbəti oxucu: {next_reader}"
            )
            if queue.is_empty():
                self.wait_queues.remove(book_id)
            return f"{previous_borrower} kitabı qaytardı. Kitab növbəti oxucu {next_reader} üçün rezerv edildi."
        book.is_available = True
        book.reserved_for = None
        self.borrowing_history.append(f"{previous_borrower} kitabı qaytardı: {book.title} (ID {book_id})")
        if queue is not None and queue.is_empty():
            self.wait_queues.remove(book_id)
        return "Kitab uğurla qaytarıldı və artıq mövcuddur."

    def add_service_request(self, description, urgent=False):
        request = ("TƏCİLİ" if urgent else "ADI", description.strip())
        if urgent:
            self.service_requests.add_front(request)
        else:
            self.service_requests.add_rear(request)
        return request

    def process_service_request(self):
        if self.service_requests.is_empty():
            return None
        return self.service_requests.remove_front()

    def list_books(self):
        if len(self.books) == 0:
            print("Kataloq boşdur.")
            return
        for book in self.books:
            print(book)

    def list_sorted_books(self):
        found = False
        for book_id in self.book_tree.inorder():
            book = self.get_book(book_id)
            if book is not None:
                print(book)
                found = True
        if not found:
            print("Kataloq boşdur.")

    def show_history(self):
        if self.borrowing_history.is_empty():
            print("Hələ tarixçə yoxdur.")
            return
        for entry in self.borrowing_history:
            print("-", entry)

    def show_recently_viewed(self):
        if self.recently_viewed.is_empty():
            print("Hələ baxılmış kitab yoxdur.")
            return
        for item in self.recently_viewed:
            print("-", item)

    def load_sample_data(self):
        # 15 nümunə kitab. ID-lər qəsdən qarışıq ardıcıllıqla verilib ki,
        # BST-yə əlavə etmə sırası avtomatik olaraq sıralanmış olmasın.
        samples = [
            (1008, "The Alchemist", "Paulo Coelho", 1988),
            (1003, "Clean Code", "Robert C. Martin", 2008),
            (1012, "Python Crash Course", "Eric Matthes", 2015),
            (1001, "1984", "George Orwell", 1949),
            (1015, "The Little Prince", "Antoine de Saint-Exupéry", 1943),
            (1006, "Animal Farm", "George Orwell", 1945),
            (1010, "Sapiens: A Brief History of Humankind", "Yuval Noah Harari", 2011),
            (1004, "The Pragmatic Programmer", "Andrew Hunt və David Thomas", 1999),
            (1013, "Deep Work", "Cal Newport", 2016),
            (1002, "The Hobbit", "J. R. R. Tolkien", 1937),
            (1011, "The Lean Startup", "Eric Ries", 2011),
            (1005, "Atomic Habits", "James Clear", 2018),
            (1009, "To Kill a Mockingbird", "Harper Lee", 1960),
            (1007, "The Great Gatsby", "F. Scott Fitzgerald", 1925),
            (1014, "Introduction to Algorithms", "Thomas H. Cormen, Charles E. Leiserson, Ronald L. Rivest və Clifford Stein", 1990),
        ]
        for book_id, title, author, year in samples:
            if self.get_book(book_id) is None:
                self._insert_book_indexes(Book(book_id, title, author, year))

    @staticmethod
    def print_menu():
        print("""
========================================
        MINI LIBRARY MANAGEMENT
========================================
1. Yeni kitab əlavə et
2. Kitabı sil
3. ID ilə kitab axtar
4. Bütün kitabları göstər
5. Kitabı götür (borrow)
6. Kitabı qaytar (return)
7. Borc tarixçəsini göstər
8. Son baxılmış kitabları göstər
9. Son kataloq əməliyyatını geri al (Stack)
10. Kitabları ID üzrə sırala (BST)
11. Xidmət sorğusu əlavə et (Deque)
12. Növbəti xidmət sorğusunu icra et (Deque)
13. Hash Table statistikasını göstər
0. Çıxış
========================================
""")

    def run(self):
        while True:
            self.print_menu()
            choice = input("Seçiminiz: ").strip()
            try:
                if choice == "1":
                    book_id = int(input("Kitab ID: "))
                    title = input("Kitabın adı: ")
                    author = input("Müəllif: ")
                    year = int(input("Nəşr ili: "))
                    print("Kitab əlavə edildi." if self.add_book(book_id, title, author, year)
                          else "Bu ID artıq mövcuddur.")

                elif choice == "2":
                    book_id = int(input("Silinəcək kitabın ID-si: "))
                    if self.remove_book(book_id):
                        print("Kitab silindi. Əməliyyat Stack-də saxlanıldı və geri alına bilər.")
                    else:
                        print("Kitab tapılmadı, borcdadır və ya gözləmə növbəsi var.")

                elif choice == "3":
                    book_id = int(input("Axtarılan kitabın ID-si: "))
                    book = self.search_book(book_id)
                    print(book if book else "Kitab tapılmadı.")

                elif choice == "4":
                    self.list_books()

                elif choice == "5":
                    book_id = int(input("Kitab ID: "))
                    borrower = input("Oxucunun adı: ")
                    print(self.borrow_book(book_id, borrower))

                elif choice == "6":
                    book_id = int(input("Qaytarılan kitabın ID-si: "))
                    print(self.return_book(book_id))

                elif choice == "7":
                    self.show_history()

                elif choice == "8":
                    self.show_recently_viewed()

                elif choice == "9":
                    print(self.undo_last_catalog_change())

                elif choice == "10":
                    self.list_sorted_books()

                elif choice == "11":
                    description = input("Sorğunun təsviri: ")
                    urgent_answer = input("Təcili? (b/x): ").strip().casefold()
                    request = self.add_service_request(description, urgent_answer in ("b", "belə", "yes", "y"))
                    print("Sorğu əlavə edildi:", request)

                elif choice == "12":
                    request = self.process_service_request()
                    print("İcra edilən sorğu:" if request else "İcra üçün sorğu yoxdur.", request or "")

                elif choice == "13":
                    print("Kitab sayı:", len(self.books))
                    print("Hash Table: ölçü =", len(self.books_by_id),
                          "| bucket sayı =", self.books_by_id.capacity,
                          "| load factor =", round(self.books_by_id.load_factor, 3),
                          "| collision hadisələri =", self.books_by_id.collision_events,
                          "| ən uzun chain =", self.books_by_id.max_chain_length())

                elif choice == "0":
                    print("Proqram bağlandı. Uğurlar!")
                    break
                else:
                    print("Yanlış seçim. Menyudan 0–13 arası seçim edin.")

            except ValueError as error:
                print("Daxil edilən məlumat yanlışdır:", error)
            except IndexError as error:
                print("Əməliyyat mümkün olmadı:", error)

            input("Davam etmək üçün Enter basın...")
```

## 11. Nəticələrin şərhi və Complexity Analysis

| Struktur / əməliyyat | Orta vaxt mürəkkəbliyi | Ən pis hal | Qeyd |
|---|---:|---:|---|
| Fixed/Dynamic Array – indekslə oxuma | `O(1)` | `O(1)` | Birbaşa indeks |
| Array – axtarış | `O(n)` | `O(n)` | Xətti axtarış |
| Array – ortada əlavə/silmə | `O(n)` | `O(n)` | Elementlər sürüşdürülür |
| Dynamic Array – sona əlavə | amortizə `O(1)` | `O(n)` | Tutum dolanda resize/copy |
| Singly Linked List – başa əlavə | `O(1)` | `O(1)` | Head göstəricisi |
| Singly Linked List – sona əlavə | `O(1)` | `O(1)` | Tail saxlanılır |
| Singly Linked List – dəyərə görə axtarış/silmə | `O(n)` | `O(n)` | Düyünlər ardıcıl yoxlanır |
| Singly Linked List – sondan silmə | `O(n)` | `O(n)` | Əvvəlki düyün tapılmalıdır |
| Doubly Linked List – sondan silmə | `O(1)` | `O(1)` | `tail.prev` mövcuddur |
| Stack – push/pop | `O(1)` | `O(1)` | LIFO |
| Queue – enqueue/dequeue | `O(1)` | `O(1)` | FIFO |
| Deque – hər iki başda əməliyyat | `O(1)` | `O(1)` | Doubly Linked List üzərində |
| Hash Table – axtarış/əlavə/silmə | gözlənilən `O(1)` | `O(n)` | Collision olduqda zəncir uzana bilər |
| BST – axtarış/əlavə/silmə | `O(log n)` | `O(n)` | Ağacın balansından asılıdır |

**Benchmark və nəzəri mürəkkəblik:** Cədvəl nəzəri vaxt mürəkkəbliyini, benchmark-lar isə konkret Colab mühitində ölçülmüş icra vaxtını göstərir. Dynamic Array-də tutumun artırılması ayrı-ayrı əlavə etmə vaxtına təsir edə bilər. Linked List-lərdə sondan silmə fərqi `tail` göstəricilərindən asılıdır. BST-də sorted insertion ağacın hündürlüyünü artıraraq əməliyyatları `O(n)`-ə yaxınlaşdıra bilər.

Hesabatda benchmark cədvəllərində alınan konkret ölçüləri qeyd edin.
